## **Overview**
Pathly will use a client-server architecture consisting of:

- React Native mobile application
- Node.js / Express backend
- MongoDB database

The mobile application will communicate with the backend through REST APIs. The backend will aggregate and normalize publicly available transit data, including scheduled and real-time information, and integrate it with mapping and geocoding services.
