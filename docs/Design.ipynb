### **Overview**
Pathly is a mobile public transportation app built with React Native and Expo for riders in the New York City and Long Island Area. It will use a Firebase backend to handle user accounts, application data, and communication with external transit and mapping services, while its database, Firestore, will store Pathly-specific data such as user profiles, favorites, and preferences. Transit information will be obtained from sources such as GTFS and GTFS-Realtime feeds and presented through the mobile interface. The system is designed as a client-server architecture so that the mobile application can interact with a centralized backend and external data sources.

### **Stack**

As a team, we initially decided to use the MERN stack for its cross-platform development support (Expo) and familiarity to all members of the team. However, one of our teammates, Naseeha, suggested using Firebase as our backend because it's a lot easier to set up. After researching into it further, we ended up choosing Firebase since it can manage our authentication, data storage, and server in a single environment compared to creating our own Node.js/Express backend with MongoDB. Additionally, we're planning on using Google APIs like Google Maps and Routes API, which are very easy to do in Firebase. Hence, Firebase reduces the amount of backend infrastructure we need to develop and maintain. Our end goal is to release our app on both Android and iOS devices, so we want to spend more time on refining core features like route navigation, map search, and real-time data.

| Layer | Technology |
|---|---|
| **frontend** | React Native + Expo |
| **backend** | Firebase |
| **database** | Cloud Firestore |
| **server-side logic** | Cloud Functions for Firebase |
| **authentication** | Firebase Authentication |
| **file storage** | Firebase Cloud Storage |

We will be programming in TypeScript over JavaScript because it is statically typed, allowing us (or the LLM) to catch errors early on. Having explicit types makes it easier to code, debug, and read, which is particularly important given the scale of our project.

### **Design Goals**
In order for our app to stand out, we would like to prioritize several design goals:

- **Accuracy** - The most common complaint we observed from our survey was inaccuracy. As such, we want our app to be as accurate as possible so that users have the best transit experience.
- **Simplicity** - It should be very easy for both new and experienced riders to use. Users should be able to intuitively access the information they want with minimal number of clicks. Everything should be displayed neatly and quickly.
- **Scalability** - As we add other transit systems and APIs, we should be able to be integrate it into our app without making major changes.

### **Database**

As stated in the proposal, our target systems include:

- MTA Bus
- MTA Subway
- NYC Ferry
- LIRR
- Nassau Inter-County Express (NICE)
- Suffolk County Transit

**GTFS**, or General Transit Feed Specification, is a standardized data format that provides a structure for public transit agencies to describe transit information. It is considered the open data standard, and all of our target systems use it. For example, it can provide **static** information such as:

- stops
- routes
- trips
- schedules
- fare information

It is updated by a variable time (i.e. hourly, quarterly).
**GTFS-RT**, or GTFS Realtime, can provide **live** information such as:

- vehicle locations
- delays
- service alerts
- elevators and escalators

### **User Authentication**

We would like to use Google OAuth to authenticate our users for ease-of-use. Upon opening the app for the first time, users will be prompted to sign in or continue as guest. While having both options introduces complexity, it is much more user-friendly this way because our app does not handle sensitive information and therefore does not need much security.

Each of us will be using Claude Pro or Codex as our LLM to code. As recommended by the professor, we will use OpenSpec to ensure that the LLM understands what to build before it starts coding.
